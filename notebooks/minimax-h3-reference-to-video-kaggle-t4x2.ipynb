### 🚀 What is this notebook?

Make videos **from your own references** with **MiniMax H3** on Kaggle's free **2× T4** GPUs, **without owning a GPU**. Give the model **pictures** (a character, a product, a place), a **video** (the motion you want) and/or an **audio clip** (a voice, music or rhythm), write a prompt, and press Generate. When setup finishes it opens a simple web page.

| Feature | Detail |
|---|---|
| **Model** | MiniMax H3 reference-to-video (int8, pruned) |
| **References** | up to **9 pictures**, **3 videos** (with their soundtracks) and **3 audio clips** |
| **GPU** | Kaggle T4 x2 (2 × 15 GB VRAM) |
| **Engine** | ComfyUI v0.34.0 (pinned commit) + the small **H3 Reference Studio** plugin (code is shown below) |
| **Speed modes** | **VDN 8-step (default)**, Turbo 8-step, Turbo 4-step, or off (20 steps, slow) |
| **Extras** | Audio is generated with the video, up to 15 s, TR/EN page, history, "where the time went" summary |
| **Safety** | Every download is SHA256-verified, the public link is password-protected, uploaded files are re-encoded |

4. Click the link in the green **"Studio is ready"** box at the end. When you are done, **download your videos first**, then stop the cell with ■

### 🖼️ How to use references
1. **Add your references** in the *References* box: pictures, a video and/or an audio clip. Each one gets a **tag** such as `<Picture 1>`, `<Video 1>` or `<Audio 1>`.
2. **Point at them in your prompt** with those tags (click a tag to insert it). Example:
   `Use <Video 1> as the motion reference: <Picture 1> performs the same dance in a sunlit plaza. Audio: upbeat street music.`
3. **Start small.** Every reference adds memory and time on a T4. Begin with one or two, **Draft** quality and a short duration, then grow.

What the page does for you: reference **videos are converted to 24 fps**, anything **longer than 15 s is cut**, videos are **shrunk** to save memory, pictures are re-encoded as clean PNGs. A prompt that mentions a tag you did not add (for example `<Picture 3>` with two pictures) is rejected with a clear message before any GPU time is spent.

## Step 3: Install ComfyUI

Downloads and verifies one **pinned ComfyUI commit** and installs its dependencies. (If needed it upgrades PyTorch from the official site for the fast kernels.)

```python
# Step 3 - Install ComfyUI (pinned commit) and its dependencies
import subprocess
import sys

def sh(cmd, cwd=None):
    print("$", " ".join(map(str, cmd)))
    subprocess.run(list(map(str, cmd)), cwd=cwd, check=True)

# Only this one commit is downloaded, and git verifies its identity.
if not (COMFY_HOME / ".git").exists():
    COMFY_HOME.mkdir(parents=True, exist_ok=True)
    sh(["git", "init", "--quiet", COMFY_HOME])
    sh(["git", "-C", COMFY_HOME, "fetch", "--depth", "1", COMFYUI_REPO, COMFYUI_COMMIT])
    sh(["git", "-C", COMFY_HOME, "checkout", "--quiet", "--detach", "FETCH_HEAD"])
head = subprocess.check_output(["git", "-C", str(COMFY_HOME), "rev-parse", "HEAD"], text=True).strip()
if head != COMFYUI_COMMIT:
    raise RuntimeError(f"ComfyUI commit mismatch: {head} != {COMFYUI_COMMIT}")
print("ComfyUI commit verified:", head[:12])

sh([sys.executable, "-m", "pip", "install", "-q", "-r", COMFY_HOME / "requirements.txt"])
# huggingface_hub 2.x conflicts with Kaggle's transformers ("huggingface-hub<2.0 required"), so it stays on 1.x.
sh([sys.executable, "-m", "pip", "install", "-q", "huggingface_hub>=1.5,<2.0", "requests", "psutil"])
subprocess.run([sys.executable, "-m", "pip", "cache", "purge"], check=False)

# ---- PyTorch cu130 for the fast int8 kernels (needs NVIDIA driver >= 580) ----
# (Kaggle's preinstalled conda torch owns the import path, so it is removed first.)
def driver_major():
    try:
        out = subprocess.run(["nvidia-smi", "--query-gpu=driver_version", "--format=csv,noheader"],
                             capture_output=True, text=True, timeout=30).stdout.strip().splitlines()
        return int(out[0].split(".")[0]) if out else 0
    except Exception:
        return 0

def torch_in_subprocess():
    r = subprocess.run([sys.executable, "-c", "import torch; print(torch.__version__ + '|' + str(torch.version.cuda))"],
                       capture_output=True, text=True, timeout=120)
    return tuple(r.stdout.strip().split("|")) if r.returncode == 0 and "|" in r.stdout else ("?", "?")

DRIVER = driver_major()
print("NVIDIA driver major version:", DRIVER)
ver, cuda = torch_in_subprocess()
print(f"Right now: torch {ver} (CUDA {cuda})")
FAST_KERNELS = cuda.startswith("13.")
if UPGRADE_TORCH_TO_CU130 and not FAST_KERNELS:
    if DRIVER < 580:
        print("⚠ Driver < 580: cu130 cannot be used. Fast kernels stay off; continuing with the Turbo/VDN modes (may be slow).")
    else:
        print("Upgrading PyTorch to cu130 (from the official pytorch.org index)...")
        subprocess.run([sys.executable, "-m", "pip", "uninstall", "-y", "torch", "torchvision", "torchaudio"],
                       capture_output=True, timeout=300)
        sh([sys.executable, "-m", "pip", "install", "-q", "--no-cache-dir", "--index-url",
            "https://download.pytorch.org/whl/cu130", "torch", "torchvision", "torchaudio"])
        ver, cuda = torch_in_subprocess()
        FAST_KERNELS = cuda.startswith("13.")
        print(f"Result: torch {ver} (CUDA {cuda})")
        if FAST_KERNELS:
            subprocess.run([sys.executable, "-m", "pip", "install", "-q", "--force-reinstall", "--no-cache-dir",
                            "comfy-kitchen"], capture_output=True, timeout=600)
        else:
            print("⚠ The upgrade had no effect; the slow path will be used.")
print("Fast kernels:", "ON" if FAST_KERNELS else "OFF (generation will be slower)")

# ---- Early check: are the packages ComfyUI needs compatible with each other? (so there are no surprises in Step 6) ----
check = subprocess.run([sys.executable, "-c",
    "import torch, transformers, huggingface_hub, safetensors, aiohttp;"
    "print('torch', torch.__version__, '| transformers', transformers.__version__, '| huggingface_hub', huggingface_hub.__version__)"],
    capture_output=True, text=True, timeout=300)
if check.returncode != 0:
    raise RuntimeError("Package conflict (ComfyUI cannot start):\n" + (check.stderr or check.stdout)[-1500:])
print("✓ Packages compatible:", check.stdout.strip())
```

```python
# Step 4-a  h3_studio/__init__.py
PLUGIN_DIR = COMFY_HOME / "custom_nodes" / "h3_studio"
PLUGIN_DIR.mkdir(parents=True, exist_ok=True)
(PLUGIN_DIR / "__init__.py").write_text(r'''"""H3 Reference Studio: a simple page for MiniMax H3 reference-to-video on ComfyUI, protected by an access key.

What this custom node does (all of it is in this file, nothing is downloaded or executed elsewhere):
  1. Access key: when the environment variable H3_STUDIO_KEY is set, every request (pages, API, websocket) must carry
     the key, otherwise the server answers 401. This matters because the Kaggle notebook exposes ComfyUI through a
     public tunnel.
  2. /h3 serves the simple page (studio.html). /h3/api/reference prepares an uploaded reference picture, video or
     audio (re-encoded, length-limited, 24 fps for video). /h3/api/build turns the page's settings into a ComfyUI API
     graph made only of ComfyUI's own MiniMax H3 nodes. The page then sends that graph to /prompt.
  3. Two small loader nodes that place the text encoder / VAE on a chosen GPU (for two-GPU machines).

License: not decided yet. This file imports ComfyUI (GPL-3.0) modules, so the author must choose a compatible
license before distributing it (see README.md).
"""
import asyncio
import hmac
import logging
import math
import os
import re
import secrets
from fractions import Fraction
from pathlib import Path
from urllib.parse import urlencode

import torch
from aiohttp import web

import comfy.sd
import comfy.utils
import folder_paths
from server import PromptServer

log = logging.getLogger("h3_studio")
HERE = Path(__file__).resolve().parent
VERSION = "0.1.0-reference"

# ---- model files (all downloaded by the notebook and verified with SHA256) ----
DIT_FILE = "minimax_h3_ref2va_pruned_int8_convrot.safetensors"   # the reference-to-video base model
TEXT_ENCODER_FILE = "qwen3vl_32b_minimax_h3_nvfp4_awq.safetensors"
VIDEO_VAE_FILE = "minimax_h3_video_vae_fp16.safetensors"
AUDIO_VAE_FILE = "minimax_h3_audio_vae_fp32.safetensors"
# Turbo LoRA variants for the reference model. Only files that are really present are offered.
TURBO_VARIANTS = {
    "turbo8_768": {"file": "minimax_h3_ref2v_turbo_8step_v1.0_768p_comfyui_bf16.safetensors", "steps": 8},
    "turbo4": {"file": "minimax_h3_ref2v_turbo_4step_v0.1_comfyui_bf16.safetensors", "steps": 4},
}

# The VDN-H3 8-step stage applied to the same int8 base through a third-party node. Offered (first in the list, so it is
# the default) only if the notebook installed it completely; otherwise the Turbo modes are used.
VDN_ID = "vdn8"
VDN_STAGE_DIR = "stage-dmd-step-250-int8_convrot_comfyui"
VDN_STEPS = 8
VDN_SAMPLER = "euler"  # the node's README asks for Euler + an 8-step "simple" schedule, and no LoRA before it

# ---- limits (the notebook may override some through environment variables) ----
MAX_MEGAPIXELS = float(os.environ.get("H3_STUDIO_MAX_MP", "0.45"))
MAX_SECONDS = float(os.environ.get("H3_STUDIO_MAX_SECONDS", "10"))
MIN_SECONDS = 2.0
MAX_PROMPT_CHARS = 6000
ASPECTS = {"landscape": (16, 9), "portrait": (9, 16), "square": (1, 1)}
QUALITY = {"draft": 0.2, "standard": 0.3, "high": 0.4}  # megapixels, 1 MP = 1024 x 1024 like ComfyUI
RESOLUTION_MULTIPLE = 32
FPS = 24
SAMPLER = "res_multistep"
NORMAL_STEPS = 20

# Reference limits (the node accepts up to 9 pictures, 3 videos with their soundtracks, 3 audios).
MAX_REF_IMAGES, MAX_REF_VIDEOS, MAX_REF_AUDIOS = 9, 3, 3
REF_VIDEO_MIN_SECONDS, REF_MAX_SECONDS = 2.0, 15.0     # the node documents 2-15 s reference videos
REF_VIDEO_MAX_MEGAPIXELS = 0.3                         # reference clips are downscaled to save memory (not measured on T4)
REF_IMAGE_MAX_SHORT_EDGE = 2048                        # the node never uses more than this
MAX_UPLOAD_BYTES = 150 * 1024 * 1024
REF_SIZES = ("match", "max")
REF_NAME = re.compile(r"^h3ref_(image|video|audio)_[0-9a-f]{12}\.(png|mp4|wav)$")
TAG = re.compile(r"<\s*(Picture|Video|Audio)\s+(\d+)\s*>", re.IGNORECASE)


# =============================== access key ===============================
ACCESS_KEY = os.environ.get("H3_STUDIO_KEY", "")
COOKIE_NAME = "h3_key"


def _key_ok(value):
    return bool(ACCESS_KEY) and bool(value) and hmac.compare_digest(value.encode("utf-8"), ACCESS_KEY.encode("utf-8"))


@web.middleware
async def access_key_middleware(request, handler):
    if not ACCESS_KEY:
        return await handler(request)
    if _key_ok(request.headers.get("X-H3-Key", "")) or _key_ok(request.cookies.get(COOKIE_NAME, "")):
        return await handler(request)
    from_query = request.query.get("key", "")
    if _key_ok(from_query):
        is_websocket = request.headers.get("Upgrade", "").lower() == "websocket"
        if request.method == "GET" and not is_websocket:
            # Remember the key in a cookie and drop it from the address bar.
            rest = {k: v for k, v in request.query.items() if k != "key"}
            location = request.path + ("?" + urlencode(rest) if rest else "")
            response = web.Response(status=302, headers={"Location": location, "Cache-Control": "no-store"})
            response.set_cookie(COOKIE_NAME, from_query, httponly=True, samesite="Lax",
                                secure=request.headers.get("X-Forwarded-Proto", "") == "https", max_age=60 * 60 * 24)
            return response
        return await handler(request)
    return web.Response(status=401, text="Unauthorized. Open the full link (with ?key=...) printed by the notebook.",
                        headers={"Cache-Control": "no-store"})


def _install_access_key():
    app = PromptServer.instance.app
    if access_key_middleware not in app.middlewares:
        app.middlewares.insert(0, access_key_middleware)  # first = outermost, guards every route
    log.warning("H3 Studio: access key %s", "ENABLED" if ACCESS_KEY else "NOT SET (open access, local use only)")


_install_access_key()


# =============================== GPU monitor (diagnostics only) ===============================
def _gpu_monitor(interval=30):
    """Every `interval` seconds, while any GPU is busy, write one line to the ComfyUI log:
    utilisation and memory per GPU. It only reads nvidia-smi; it never touches the network."""
    import subprocess
    import time
    command = ["nvidia-smi", "--query-gpu=index,utilization.gpu,memory.used,memory.total",
               "--format=csv,noheader,nounits"]
    while True:
        time.sleep(interval)
        try:
            rows = [r.split(",") for r in subprocess.run(command, capture_output=True, text=True,
                                                         timeout=10).stdout.strip().splitlines()]
            if any(int(r[1]) >= 5 for r in rows):
                log.info("h3-monitor " + " | ".join("gpu%s %s%% %.1f/%.1f GiB" % (
                    r[0].strip(), r[1].strip(), int(r[2]) / 1024, int(r[3]) / 1024) for r in rows))
        except Exception:
            pass


if os.environ.get("H3_STUDIO_MONITOR", "1") != "0":
    import threading
    threading.Thread(target=_gpu_monitor, name="h3-gpu-monitor", daemon=True).start()


# =============================== two-device loaders ===============================
def _devices():
    return ["cuda:%d" % i for i in range(torch.cuda.device_count())] + ["cpu"]


class H3StudioTextEncoder:
    """CLIP loader with an explicit device, so a 15 GB text encoder can live on the second GPU."""

    @classmethod
    def INPUT_TYPES(cls):
        return {"required": {"clip_name": (folder_paths.get_filename_list("text_encoders"),),
                             "load_to": (_devices(),)}}

    RETURN_TYPES = ("CLIP",)
    FUNCTION = "load"
    CATEGORY = "H3 Studio"

    def load(self, clip_name, load_to):
        clip_type = getattr(comfy.sd.CLIPType, "MINIMAX", None)
        if clip_type is None:
            raise RuntimeError("This ComfyUI is too old for MiniMax H3.")
        path = folder_paths.get_full_path_or_raise("text_encoders", clip_name)
        options = {"load_device": torch.device(load_to), "offload_device": torch.device("cpu")}
        clip = comfy.sd.load_clip(ckpt_paths=[path], embedding_directory=folder_paths.get_folder_paths("embeddings"),
                                  clip_type=clip_type, model_options=options)
        return (clip,)


class H3StudioVAELoader:
    """VAE loader with an explicit device."""

    @classmethod
    def INPUT_TYPES(cls):
        return {"required": {"vae_name": (folder_paths.get_filename_list("vae"),), "load_to": (_devices(),)}}

    RETURN_TYPES = ("VAE",)
    FUNCTION = "load"
    CATEGORY = "H3 Studio"

    def load(self, vae_name, load_to):
        path = folder_paths.get_full_path_or_raise("vae", vae_name)
        sd, metadata = comfy.utils.load_torch_file(path, return_metadata=True)
        vae = comfy.sd.VAE(sd=sd, device=torch.device(load_to), metadata=metadata)
        vae.throw_exception_if_invalid()
        return (vae,)


NODE_CLASS_MAPPINGS = {"H3StudioTextEncoder": H3StudioTextEncoder, "H3StudioVAELoader": H3StudioVAELoader}
NODE_DISPLAY_NAME_MAPPINGS = {"H3StudioTextEncoder": "H3 Studio Text Encoder (choose GPU)",
                              "H3StudioVAELoader": "H3 Studio VAE Loader (choose GPU)"}


# =============================== reference media preparation ===============================
def _input_dir():
    return folder_paths.get_input_directory()


def _new_name(kind, extension):
    return "h3ref_%s_%s.%s" % (kind, secrets.token_hex(6), extension)


def prepare_image(source):
    """Re-encode a picture as a clean RGB PNG (drops metadata, handles rotation and transparency)."""
    from PIL import Image, ImageOps
    try:
        with Image.open(source) as opened:
            opened.load()
            picture = ImageOps.exif_transpose(opened)
            if picture.mode in ("RGBA", "LA") or "transparency" in picture.info:
                background = Image.new("RGB", picture.size, (255, 255, 255))
                background.paste(picture.convert("RGBA"), mask=picture.convert("RGBA").split()[-1])
                picture = background
            else:
                picture = picture.convert("RGB")
    except Exception:
        raise ValueError("This file is not a picture I can read (use PNG, JPG or WebP).")
    width, height = picture.size
    shortest = min(width, height)
    if shortest > REF_IMAGE_MAX_SHORT_EDGE:
        scale = REF_IMAGE_MAX_SHORT_EDGE / shortest
        picture = picture.resize((max(1, round(width * scale)), max(1, round(height * scale))), Image.LANCZOS)
    name = _new_name("image", "png")
    picture.save(os.path.join(_input_dir(), name), format="PNG")
    return {"kind": "image", "name": name, "width": picture.size[0], "height": picture.size[1]}


def _write_audio(source, destination, max_seconds):
    """Decode the first audio track, convert to 44.1 kHz stereo 16-bit WAV, keep at most max_seconds.
    Returns the length in seconds, or None if the file has no audio track."""
    import av
    try:
        container = av.open(source)
    except Exception:
        raise ValueError("This file is not audio I can read (use WAV, MP3, M4A, FLAC or OGG).")
    with container:
        stream = next((s for s in container.streams if s.type == "audio"), None)
        if stream is None:
            return None
        resampler = av.AudioResampler(format="s16", layout="stereo", rate=44100)
        output = av.open(destination, "w", format="wav")
        encoder = output.add_stream("pcm_s16le", rate=44100)
        encoder.layout = "stereo"
        limit, written = int(max_seconds * 44100), 0
        try:
            def push(frames):
                nonlocal written
                for frame in frames or []:
                    for packet in encoder.encode(frame):
                        output.mux(packet)
                    written += frame.samples
            for decoded in container.decode(stream):
                push(resampler.resample(decoded))
                if written >= limit:
                    break
            if written < limit:
                push(resampler.resample(None))
            for packet in encoder.encode(None):
                output.mux(packet)
        finally:
            output.close()
    return written / 44100.0


def prepare_audio(source):
    name = _new_name("audio", "wav")
    path = os.path.join(_input_dir(), name)
    seconds = _write_audio(source, path, REF_MAX_SECONDS)
    if seconds is None:
        raise ValueError("This file has no audio track.")
    if seconds < 0.5:
        os.unlink(path)
        raise ValueError("This audio is too short (minimum 0.5 seconds).")
    return {"kind": "audio", "name": name, "seconds": round(seconds, 2),
            "trimmed": seconds >= REF_MAX_SECONDS - 0.05}


def prepare_video(source):
    """Re-encode a reference video as H.264 at exactly 24 fps (the model assumes 24 fps), at most 15 s and ~0.3 MP,
    and extract its soundtrack (if any) as a separate WAV."""
    import av
    try:
        container = av.open(source)
    except Exception:
        raise ValueError("This file is not a video I can read (use MP4, MOV or WEBM).")
    with container:
        stream = next((s for s in container.streams if s.type == "video"), None)
        if stream is None:
            raise ValueError("This file has no video track.")
        width, height = stream.codec_context.width, stream.codec_context.height
        if not width or not height:
            raise ValueError("Could not read the video size.")
        scale = min(1.0, math.sqrt(REF_VIDEO_MAX_MEGAPIXELS * 1024 * 1024 / (width * height)))
        target_w, target_h = max(32, int(width * scale) // 2 * 2), max(32, int(height * scale) // 2 * 2)
        name = _new_name("video", "mp4")
        path = os.path.join(_input_dir(), name)
        output = av.open(path, "w")
        encoder = None
        for codec in ("libx264", "h264", "mpeg4"):
            try:
                encoder = output.add_stream(codec, rate=FPS)
                break
            except Exception:
                encoder = None
        if encoder is None:
            output.close()
            raise ValueError("This server cannot encode video.")
        encoder.width, encoder.height, encoder.pix_fmt = target_w, target_h, "yuv420p"
        if encoder.codec_context.name == "libx264":
            encoder.options = {"crf": "18", "preset": "veryfast"}
        limit, index, current, first_time = int(REF_MAX_SECONDS * FPS), 0, None, None

        def emit():
            nonlocal index
            frame = current.reformat(width=target_w, height=target_h, format="yuv420p")
            frame.pts, frame.time_base = index, Fraction(1, FPS)
            for packet in encoder.encode(frame):
                output.mux(packet)
            index += 1

        try:
            for decoded in container.decode(stream):
                if decoded.pts is None:
                    continue
                moment = float(decoded.pts * stream.time_base)
                first_time = moment if first_time is None else first_time
                moment -= first_time
                while current is not None and index / FPS < moment and index < limit:
                    emit()
                current = decoded
                if index >= limit:
                    break
            if current is not None and index < limit:
                emit()  # hold the last picture for one frame
            for packet in encoder.encode(None):
                output.mux(packet)
        finally:
            output.close()
    seconds = index / FPS
    if seconds < REF_VIDEO_MIN_SECONDS:
        os.unlink(path)
        raise ValueError("Reference videos must be at least %g seconds long (this one is %.1f s)."
                         % (REF_VIDEO_MIN_SECONDS, seconds))
    audio_name = None
    audio_path = os.path.join(_input_dir(), _new_name("audio", "wav"))
    try:
        if _write_audio(source, audio_path, min(seconds, REF_MAX_SECONDS)) is not None:
            audio_name = os.path.basename(audio_path)
    except Exception:
        audio_name = None  # a broken soundtrack is not fatal: the video is still usable without it
        if os.path.exists(audio_path):
            os.unlink(audio_path)
    return {"kind": "video", "name": name, "seconds": round(seconds, 2), "width": target_w, "height": target_h,
            "audio": audio_name, "trimmed": seconds >= REF_MAX_SECONDS - 0.05}


PREPARE = {"image": prepare_image, "video": prepare_video, "audio": prepare_audio}


# =============================== graph builder ===============================
def resolution(aspect, megapixels):
    """Same maths as ComfyUI's ResolutionSelector node (1 MP = 1024 x 1024), rounded to a multiple of 32."""
    wr, hr = ASPECTS[aspect]
    scale = math.sqrt(megapixels * 1024 * 1024 / (wr * hr))
    return (round(wr * scale / RESOLUTION_MULTIPLE) * RESOLUTION_MULTIPLE,
            round(hr * scale / RESOLUTION_MULTIPLE) * RESOLUTION_MULTIPLE)


def aligned_frames(seconds):
    """H3's video VAE needs 17*n+5 frames."""
    frames = max(5, round(seconds * FPS))
    return frames + (5 - frames % 17) % 17


def _have(folder, name):
    try:
        return name in folder_paths.get_filename_list(folder)
    except Exception:
        return False


VDN_REQUIRED_FILES = ("model_spec.json", "linear_branch/model_int8_convrot_comfyui.safetensors",
                      "adapters/default/adapter_model.safetensors", "adapters/turbo/adapter_model.safetensors")


def vdn_available():
    """True only if the VDN node is loaded and every stage file and the AdaLN grid are on disk."""
    try:
        import nodes
        if "VDNH3Fast" not in nodes.NODE_CLASS_MAPPINGS:
            return False
    except Exception:
        return False
    base = os.path.join(folder_paths.models_dir, "vdn")
    return (os.path.isfile(os.path.join(base, "h3_silu_temb_grid.safetensors"))
            and all(os.path.isfile(os.path.join(base, VDN_STAGE_DIR, *rel.split("/"))) for rel in VDN_REQUIRED_FILES))


def turbo_options():
    """Speed modes that are really installed, in display order. VDN (when installed) comes first = the default."""
    options = [{"id": key, "steps": spec["steps"]} for key, spec in TURBO_VARIANTS.items()
               if _have("loras", spec["file"])]
    if vdn_available():
        options.insert(0, {"id": VDN_ID, "steps": VDN_STEPS})
    return options


def missing_files():
    needed = [("diffusion_models", DIT_FILE), ("text_encoders", TEXT_ENCODER_FILE),
              ("vae", VIDEO_VAE_FILE), ("vae", AUDIO_VAE_FILE)]
    return [name for folder, name in needed if not _have(folder, name)]


def _ref_name(value, kind, label):
    if not isinstance(value, str) or not REF_NAME.match(value) or not value.startswith("h3ref_%s_" % kind):
        raise ValueError("Invalid %s reference. Upload it again." % label)
    if not os.path.isfile(os.path.join(_input_dir(), value)):
        raise ValueError("A %s reference is no longer on the server. Upload it again." % label)
    return value


def _clean_refs(refs):
    """Validate the references sent by the page. Returns (images, videos, audios, image_size)."""
    if refs in (None, ""):
        refs = {}
    if not isinstance(refs, dict):
        raise ValueError("Invalid references.")
    images = [_ref_name(n, "image", "picture") for n in (refs.get("images") or [])]
    audios = [_ref_name(n, "audio", "audio") for n in (refs.get("audios") or [])]
    videos = []
    for item in refs.get("videos") or []:
        if not isinstance(item, dict):
            raise ValueError("Invalid video reference.")
        entry = {"name": _ref_name(item.get("name"), "video", "video"), "audio": None}
        if item.get("soundtrack") and item.get("audio"):
            entry["audio"] = _ref_name(item.get("audio"), "audio", "soundtrack")
        videos.append(entry)
    if len(images) > MAX_REF_IMAGES or len(videos) > MAX_REF_VIDEOS or len(audios) > MAX_REF_AUDIOS:
        raise ValueError("Too many references (max %d pictures, %d videos, %d audios)."
                         % (MAX_REF_IMAGES, MAX_REF_VIDEOS, MAX_REF_AUDIOS))
    size = refs.get("image_size", "match")
    if size not in REF_SIZES:
        raise ValueError("Unknown reference detail option.")
    return images, videos, audios, size


def _check_tags(prompt, n_images, n_videos, n_audios):
    """The prompt may only mention references that exist: <Picture 2> needs two pictures, and so on."""
    have = {"picture": n_images, "video": n_videos, "audio": n_audios}
    for kind, number in TAG.findall(prompt):
        kind, number = kind.lower(), int(number)
        if number < 1 or number > have[kind]:
            raise ValueError("Your prompt mentions <%s %d> but you added %d %s reference%s." % (
                kind.capitalize(), number, have[kind], kind, "" if have[kind] == 1 else "s"))


def build_prompt(data):
    """Validate the page's settings and return (ComfyUI API graph, summary)."""
    if not isinstance(data, dict):
        raise ValueError("Invalid request.")
    prompt = data.get("prompt")
    if not isinstance(prompt, str) or not prompt.strip():
        raise ValueError("Write a prompt first.")
    if len(prompt) > MAX_PROMPT_CHARS:
        raise ValueError("Prompt is too long (max %d characters)." % MAX_PROMPT_CHARS)
    aspect = data.get("aspect", "landscape")
    if aspect not in ASPECTS:
        raise ValueError("Unknown aspect ratio.")
    quality = data.get("quality", "draft")
    if quality not in QUALITY:
        raise ValueError("Unknown quality preset.")
    megapixels = QUALITY[quality]
    if megapixels > MAX_MEGAPIXELS:
        raise ValueError("This quality is above the limit set for this session.")
    seconds = data.get("seconds", 5)
    if isinstance(seconds, bool) or not isinstance(seconds, (int, float)) or not math.isfinite(seconds):
        raise ValueError("Duration must be a number of seconds.")
    if not MIN_SECONDS <= seconds <= MAX_SECONDS:
        raise ValueError("Duration must be between %g and %g seconds." % (MIN_SECONDS, MAX_SECONDS))
    seed = data.get("seed")
    if type(seed) is not int or not 0 <= seed <= 2 ** 53 - 1:
        raise ValueError("Seed must be an integer.")
    images, videos, audios, image_size = _clean_refs(data.get("refs"))
    soundtracks = sum(1 for video in videos if video["audio"])
    _check_tags(prompt, len(images), len(videos), soundtracks + len(audios))

    turbo = data.get("turbo", "auto")  # "off", a variant id, or "auto"
    available = turbo_options()
    if turbo in (False, "off", None, ""):
        variant = None
    elif turbo in (True, "auto"):
        variant = available[0]["id"] if available else None
    elif isinstance(turbo, str) and (turbo in TURBO_VARIANTS or turbo == VDN_ID):
        variant = turbo
    else:
        raise ValueError("Unknown speed mode.")

    absent = missing_files()
    if absent:
        raise ValueError("Model files are missing: " + ", ".join(absent))
    if variant and not any(option["id"] == variant for option in available):
        raise ValueError("That speed mode is not installed; choose another mode or re-run the download cell.")

    width, height = resolution(aspect, megapixels)
    frames = aligned_frames(seconds)
    steps = VDN_STEPS if variant == VDN_ID else TURBO_VARIANTS[variant]["steps"] if variant else NORMAL_STEPS

    graph = {}

    def node(name, class_type, **inputs):
        graph[name] = {"class_type": class_type, "inputs": inputs}
        return [name, 0]

    gpus = torch.cuda.device_count()
    node("unet", "UNETLoader", unet_name=DIT_FILE, weight_dtype="default")
    model = ["unet", 0]
    sampler_name = SAMPLER
    if variant == VDN_ID:  # the VDN patch replaces the plain model; no LoRA may sit before it
        model = node("vdn", "VDNH3Fast", model=["unet", 0], vdn_checkpoint=VDN_STAGE_DIR)
        sampler_name = VDN_SAMPLER
    elif variant:
        model = node("lora", "LoraLoaderModelOnly", model=["unet", 0], lora_name=TURBO_VARIANTS[variant]["file"],
                     strength_model=1.0)
    if gpus >= 2:  # text encoder alone on GPU 1, video/audio VAE with the diffusion model on GPU 0
        clip = node("clip", "H3StudioTextEncoder", clip_name=TEXT_ENCODER_FILE, load_to="cuda:1")
        vae_v = node("vae_v", "H3StudioVAELoader", vae_name=VIDEO_VAE_FILE, load_to="cuda:0")
        vae_a = node("vae_a", "H3StudioVAELoader", vae_name=AUDIO_VAE_FILE, load_to="cuda:0")
    else:
        clip = node("clip", "CLIPLoader", clip_name=TEXT_ENCODER_FILE, type="minimax", device="default")
        vae_v = node("vae_v", "VAELoader", vae_name=VIDEO_VAE_FILE)
        vae_a = node("vae_a", "VAELoader", vae_name=AUDIO_VAE_FILE)

    # References. The conditioning node names its dynamic inputs "<group>.<prefix><index>".
    ref_inputs = {}
    for index, name in enumerate(images):
        ref_inputs["ref_images.ref_image_%d" % index] = node("img%d" % index, "LoadImage", image=name)
    for index, video in enumerate(videos):
        loaded = node("vid%d" % index, "LoadVideo", file=video["name"])
        ref_inputs["ref_videos.ref_video_%d" % index] = node(
            "vidc%d" % index, "GetVideoComponents", video=loaded)  # output 0 = the frames (already 24 fps)
        if video["audio"]:
            ref_inputs["ref_video_audios.ref_video_audio_%d" % index] = node(
                "vaud%d" % index, "LoadAudio", audio=video["audio"])
    for index, name in enumerate(audios):
        ref_inputs["ref_audios.ref_audio_%d" % index] = node("aud%d" % index, "LoadAudio", audio=name)

    node("cond", "MiniMaxH3ReferenceToVideo", clip=clip, vae=vae_v, audio_vae=vae_a, prompt=prompt.strip(),
         width=width, height=height, length=frames, ref_image_size=image_size, **ref_inputs)
    noise = node("noise", "RandomNoise", noise_seed=seed)
    guider = node("guider", "BasicGuider", model=model, conditioning=["cond", 0])
    sampler = node("sampler", "KSamplerSelect", sampler_name=sampler_name)
    sigmas = node("sched", "BasicScheduler", model=model, scheduler="simple", steps=steps, denoise=1.0)
    node("sample", "SamplerCustomAdvanced", noise=noise, guider=guider, sampler=sampler, sigmas=sigmas,
         latent_image=["cond", 1])
    picture = node("dec_v", "VAEDecode", samples=["sample", 0], vae=vae_v)
    sound = node("dec_a", "VAEDecodeAudio", samples=["sample", 0], vae=vae_a)
    video = node("video", "CreateVideo", images=picture, fps=FPS, audio=sound)
    node("save", "SaveVideo", video=video, filename_prefix="video/H3Studio", format="auto", codec="auto")

    summary = {"width": width, "height": height, "frames": frames, "seconds": round(frames / FPS, 3),
               "steps": steps, "turbo": variant or "off", "gpus": gpus, "seed": seed,
               "references": {"pictures": len(images), "videos": len(videos), "audios": len(audios) + soundtracks}}
    return graph, summary


def public_config():
    count = torch.cuda.device_count()
    return {"version": VERSION, "edition": "reference", "gpus": [torch.cuda.get_device_name(i) for i in range(count)],
            "max_megapixels": MAX_MEGAPIXELS, "min_seconds": MIN_SECONDS, "max_seconds": MAX_SECONDS,
            "quality": {name: dict(zip(("width", "height"), resolution("landscape", mp)), megapixels=mp)
                        for name, mp in QUALITY.items() if mp <= MAX_MEGAPIXELS},
            "aspects": list(ASPECTS), "multiple": RESOLUTION_MULTIPLE, "fps": FPS,
            "turbo_options": turbo_options(), "normal_steps": NORMAL_STEPS, "missing": missing_files(),
            "references": {"max_images": MAX_REF_IMAGES, "max_videos": MAX_REF_VIDEOS, "max_audios": MAX_REF_AUDIOS,
                           "video_min_seconds": REF_VIDEO_MIN_SECONDS, "max_seconds": REF_MAX_SECONDS,
                           "max_upload_mb": MAX_UPLOAD_BYTES // (1024 * 1024)}}


# =============================== routes ===============================
routes = PromptServer.instance.routes


@routes.get("/h3")
async def h3_page(request):
    page = (HERE / "studio.html").read_text(encoding="utf-8")
    return web.Response(text=page, content_type="text/html", headers={"Cache-Control": "no-store"})


@routes.get("/h3/api/config")
async def h3_config(request):
    return web.json_response(public_config(), headers={"Cache-Control": "no-store"})


@routes.post("/h3/api/reference")
async def h3_reference(request):
    """Receive one picture/video/audio (multipart: field "kind" first, then field "file") and prepare it."""
    kind, temporary = None, None
    try:
        reader = await request.multipart()
        while True:
            part = await reader.next()
            if part is None:
                break
            if part.name == "kind":
                kind = (await part.text()).strip().lower()
            elif part.name == "file":
                if kind not in PREPARE:
                    return web.json_response({"error": "Unknown reference type."}, status=400)
                temporary = os.path.join(folder_paths.get_temp_directory(), "h3upload_" + secrets.token_hex(8))
                os.makedirs(os.path.dirname(temporary), exist_ok=True)
                size = 0
                with open(temporary, "wb") as handle:
                    while True:
                        chunk = await part.read_chunk(1024 * 1024)
                        if not chunk:
                            break
                        size += len(chunk)
                        if size > MAX_UPLOAD_BYTES:
                            return web.json_response(
                                {"error": "File is too large (max %d MB)." % (MAX_UPLOAD_BYTES // (1024 * 1024))},
                                status=413)
                        handle.write(chunk)
        if temporary is None or kind not in PREPARE:
            return web.json_response({"error": "No file received."}, status=400)
        loop = asyncio.get_running_loop()
        info = await loop.run_in_executor(None, PREPARE[kind], temporary)
        return web.json_response(info)
    except ValueError as error:
        return web.json_response({"error": str(error)}, status=400)
    except Exception:
        log.exception("reference upload failed")
        return web.json_response({"error": "Could not process this file."}, status=500)
    finally:
        if temporary and os.path.exists(temporary):
            os.unlink(temporary)


@routes.post("/h3/api/build")
async def h3_build(request):
    try:
        data = await request.json()
    except Exception:
        return web.json_response({"error": "Invalid request."}, status=400)
    try:
        graph, summary = build_prompt(data)
    except ValueError as error:
        return web.json_response({"error": str(error)}, status=400)
    return web.json_response({"prompt": graph, "summary": summary})
''', encoding="utf-8")
print("written:", PLUGIN_DIR / "__init__.py")
```

```python
# Step 4-b  h3_studio/studio.html
PLUGIN_DIR = COMFY_HOME / "custom_nodes" / "h3_studio"
PLUGIN_DIR.mkdir(parents=True, exist_ok=True)
(PLUGIN_DIR / "studio.html").write_text(r'''<!doctype html>
<html lang="en">
<head>
<meta charset="utf-8">
<meta name="viewport" content="width=device-width, initial-scale=1">
<title>H3 Reference Studio</title>
<style>
  :root {
    --bg: #0f141b; --panel: #171e28; --panel2: #1d2633; --line: #2a3546; --text: #e8edf4; --muted: #9aa7b8;
    --accent: #5db3ff; --accent-ink: #06121f; --good: #46d39a; --bad: #ff7a7a; --warn: #f2c14e;
  }
  @media (prefers-color-scheme: light) {
    :root { --bg: #f3f5f8; --panel: #fff; --panel2: #eef2f7; --line: #d5dce6; --text: #16202c; --muted: #5b6878;
            --accent: #1769c2; --accent-ink: #fff; --good: #138a5a; --bad: #c0392b; --warn: #8a6100; }
  }
  * { box-sizing: border-box; }
  body { margin: 0; background: var(--bg); color: var(--text); font: 15px/1.5 system-ui, "Segoe UI", sans-serif; }
  header { display: flex; align-items: center; justify-content: space-between; gap: 12px; padding: 14px 20px;
           border-bottom: 1px solid var(--line); }
  header h1 { margin: 0; font-size: 18px; letter-spacing: .02em; }
  header .right { display: flex; gap: 8px; align-items: center; }
  a { color: var(--accent); }
  main { display: grid; grid-template-columns: minmax(300px, 440px) 1fr; gap: 16px; padding: 16px 20px; max-width: 1280px; margin: 0 auto; }
  @media (max-width: 860px) { main { grid-template-columns: 1fr; padding: 12px 16px; } }
  .card { background: var(--panel); border: 1px solid var(--line); border-radius: 12px; padding: 16px; }
  label.field { display: block; margin: 14px 0 6px; font-weight: 600; font-size: 13px; color: var(--muted); }
  textarea, select, input[type=number], input[type=text] { width: 100%; background: var(--panel2); color: var(--text);
    border: 1px solid var(--line); border-radius: 8px; padding: 9px 10px; font: inherit; }
  textarea { min-height: 150px; resize: vertical; }
  .seg { display: flex; gap: 6px; flex-wrap: wrap; }
  .seg button { flex: 1; min-width: 84px; background: var(--panel2); color: var(--text); border: 1px solid var(--line);
    border-radius: 8px; padding: 8px 6px; cursor: pointer; font: inherit; }
  .seg button.on { border-color: var(--accent); box-shadow: inset 0 0 0 1px var(--accent); }
  .seg small { display: block; color: var(--muted); font-size: 11px; }
  .row { display: flex; gap: 10px; align-items: center; }
  .row > * { flex: 1; }
  .hint { color: var(--muted); font-size: 12.5px; margin-top: 4px; }
  .primary { width: 100%; margin-top: 16px; padding: 12px; border: 0; border-radius: 10px; background: var(--accent);
    color: var(--accent-ink); font: 700 16px system-ui; cursor: pointer; }
  .primary:disabled { opacity: .55; cursor: not-allowed; }
  .ghost { background: transparent; color: var(--text); border: 1px solid var(--line); border-radius: 8px; padding: 6px 10px;
    cursor: pointer; font: inherit; }
  .stage { aspect-ratio: 16 / 9; background: #000; border-radius: 10px; display: flex; align-items: center; justify-content: center;
    overflow: hidden; position: relative; }
  .stage video { width: 100%; height: 100%; object-fit: contain; background: #000; }
  .overlay { position: absolute; inset: 0; display: flex; flex-direction: column; align-items: center; justify-content: center;
    gap: 8px; padding: 18px; text-align: center; color: #dfe8f3; background: rgba(8, 12, 18, .78); }
  .overlay .big { font-size: 28px; font-weight: 700; }
  .bar { width: min(420px, 80%); height: 8px; background: #2a3546; border-radius: 99px; overflow: hidden; }
  .bar i { display: block; height: 100%; width: 0%; background: var(--accent); transition: width .3s; }
  .muted { color: var(--muted); }
  .msg { margin-top: 12px; padding: 10px 12px; border-radius: 8px; border: 1px solid var(--line); background: var(--panel2); font-size: 13.5px; }
  .msg.err { border-color: var(--bad); color: var(--bad); }
  .msg.warn { border-color: var(--warn); color: var(--warn); }
  .msg.ok { border-color: var(--good); color: var(--good); }
  .tools { display: flex; gap: 8px; margin-top: 10px; flex-wrap: wrap; align-items: center; }
  .history { margin-top: 14px; display: flex; flex-direction: column; gap: 6px; }
  .history button { text-align: left; background: var(--panel2); color: var(--text); border: 1px solid var(--line); border-radius: 8px;
    padding: 7px 10px; cursor: pointer; font: inherit; }
  .examples button { background: var(--panel2); color: var(--muted); border: 1px solid var(--line); border-radius: 99px;
    padding: 4px 10px; cursor: pointer; font-size: 12.5px; margin: 4px 6px 0 0; }
  /* references */
  .refs { margin-top: 14px; border: 1px solid var(--line); border-radius: 10px; padding: 10px 12px; background: var(--panel2); }
  .refs h3 { margin: 0 0 4px; font-size: 14px; }
  .refgroup { margin-top: 12px; }
  .refgroup > .head { display: flex; align-items: center; justify-content: space-between; gap: 8px; font-weight: 600; font-size: 13px; }
  .refitem { display: flex; align-items: center; gap: 10px; margin-top: 8px; padding: 6px; border: 1px solid var(--line);
    border-radius: 8px; background: var(--panel); flex-wrap: wrap; }
  .refitem img, .refitem video { width: 64px; height: 48px; object-fit: cover; border-radius: 6px; background: #000; }
  .refitem audio { height: 32px; width: 170px; }
  .refitem .meta { flex: 1; min-width: 120px; font-size: 12.5px; }
  .tag { background: var(--accent); color: var(--accent-ink); border: 0; border-radius: 6px; padding: 2px 8px; margin: 2px 4px 2px 0;
    font: 600 12px ui-monospace, Consolas, monospace; cursor: pointer; }
  .tag.audio { background: #b58cff; }
  .refstatus { min-height: 20px; font-size: 12.5px; margin-top: 8px; }
  footer { max-width: 1280px; margin: 0 auto; padding: 4px 20px 28px; color: var(--muted); font-size: 12.5px; }
  .brand { display: flex; flex-wrap: wrap; align-items: center; justify-content: center; gap: 10px; padding: 10px 16px;
           background: linear-gradient(135deg, #0f0c29 0%, #302b63 50%, #24243e 100%); color: #dcdcff; font-size: 13.5px; }
  .brand b { color: #fff; }
  .soc { padding: 5px 14px; border-radius: 99px; color: #fff !important; text-decoration: none; font-weight: 600; font-size: 12.5px; }
  .soc:hover { filter: brightness(1.12); }
  .yt { background: linear-gradient(135deg, #ff0000, #cc0000); }
  .ig { background: linear-gradient(135deg, #833ab4, #fd1d1d 50%, #f77737); }
  .tt { background: linear-gradient(135deg, #000, #333); border: 1px solid #444; }
  .links { margin-top: 8px; }
  .links a { margin-right: 10px; }
</style>
</head>
<body>
<header>
  <h1>H3 Reference Studio <span class="muted" data-i="tag"></span></h1>
  <div class="right">
    <a href="/" target="_blank" rel="noopener" data-i="advanced"></a>
    <button class="ghost" id="langBtn">TR / EN</button>
  </div>
</header>

<div class="brand">
  <span>🎬 Created by <b>TheBuildAI</b></span>
  <a class="soc yt" href="https://www.youtube.com/@thebuildai?sub_confirmation=1" target="_blank" rel="noopener">▶ Subscribe on YouTube</a>
  <a class="soc ig" href="https://www.instagram.com/thebuildai/" target="_blank" rel="noopener">📷 Instagram</a>
  <a class="soc tt" href="https://www.tiktok.com/@the.build.ai" target="_blank" rel="noopener">🎵 TikTok</a>
</div>

<main>
  <section class="card" aria-label="settings">
    <label class="field" for="prompt" data-i="describe"></label>
    <textarea id="prompt" data-ip="promptPh"></textarea>
    <div class="examples" id="examples"></div>
    <div class="hint" data-i="promptTip"></div>

    <div class="refs">
      <h3 data-i="refsTitle"></h3>
      <div class="hint" data-i="refsHint"></div>

      <div class="refgroup">
        <div class="head"><span><span data-i="pictures"></span> <span class="muted" id="countImages"></span></span>
          <button class="ghost" id="addImage" type="button" data-i="addPictures"></button></div>
        <input type="file" id="fileImage" accept="image/png,image/jpeg,image/webp" multiple hidden>
        <div id="listImages"></div>
      </div>

      <div class="refgroup">
        <div class="head"><span><span data-i="videos"></span> <span class="muted" id="countVideos"></span></span>
          <button class="ghost" id="addVideo" type="button" data-i="addVideo"></button></div>
        <input type="file" id="fileVideo" accept="video/mp4,video/quicktime,video/webm,video/x-matroska,.mp4,.mov,.webm,.mkv" hidden>
        <div id="listVideos"></div>
      </div>

      <div class="refgroup">
        <div class="head"><span><span data-i="audios"></span> <span class="muted" id="countAudios"></span></span>
          <button class="ghost" id="addAudio" type="button" data-i="addAudio"></button></div>
        <input type="file" id="fileAudio" accept="audio/*,.wav,.mp3,.m4a,.flac,.ogg" hidden>
        <div id="listAudios"></div>
      </div>

      <div class="refstatus" id="refStatus"></div>

      <label class="field" for="imageSize" data-i="detail"></label>
      <select id="imageSize"></select>
      <div class="hint" data-i="detailHint"></div>
    </div>

    <label class="field" data-i="frame"></label>
    <div class="seg" id="aspect">
      <button data-v="landscape" class="on"><span data-i="landscape"></span><small>16:9</small></button>
      <button data-v="portrait"><span data-i="portrait"></span><small>9:16</small></button>
      <button data-v="square"><span data-i="square"></span><small>1:1</small></button>
    </div>

    <label class="field" data-i="quality"></label>
    <div class="seg" id="quality"></div>
    <div class="hint" id="resText"></div>

    <label class="field" for="seconds"><span data-i="duration"></span>: <b id="secLabel"></b></label>
    <input type="range" id="seconds" style="width:100%">
    <div class="hint" id="secHint"></div>

    <label class="field" for="turbo" data-i="turbo"></label>
    <select id="turbo"></select>
    <div class="hint" data-i="turboHint"></div>

    <label class="field" for="seed">Seed</label>
    <div class="row">
      <input type="number" id="seed" min="0" step="1">
      <button class="ghost" id="dice" data-i="random"></button>
    </div>

    <button class="primary" id="go" data-i="generate"></button>
    <button class="ghost" id="stop" style="width:100%;margin-top:8px;display:none" data-i="cancel"></button>
    <div id="notice"></div>
  </section>

  <section class="card" aria-label="result">
    <div class="stage" id="stage">
      <video id="player" controls playsinline hidden></video>
      <div class="overlay" id="overlay"><div class="big" data-i="ready"></div><div class="muted" data-i="readyHint"></div></div>
    </div>
    <div class="tools" id="tools" hidden>
      <a class="ghost" id="download" download data-i="download"></a>
      <span class="muted" id="resultInfo"></span>
    </div>
    <div class="hint" id="breakdown" style="margin-top:10px"></div>
    <div class="msg warn" data-i="ephemeral"></div>
    <div class="history" id="history"></div>
  </section>
</main>
<footer>
  <span data-i="license"></span>
  <span class="muted" id="gpuInfo"></span>
  <div class="links">Free &amp; open source · Built by <b>TheBuildAI</b> ·
    <a href="https://www.youtube.com/@thebuildai" target="_blank" rel="noopener">YouTube</a>
    <a href="https://www.instagram.com/thebuildai/" target="_blank" rel="noopener">Instagram</a>
    <a href="https://www.tiktok.com/@the.build.ai" target="_blank" rel="noopener">TikTok</a>
    <a href="https://github.com/cafermutluozkan" target="_blank" rel="noopener">GitHub</a>
    <a href="https://www.thebuildai.tech/" target="_blank" rel="noopener">thebuildai.tech</a></div>
</footer>

<script>
"use strict";
const T = {
  en: {
    tag: "· Reference to video, MiniMax H3 on Kaggle", advanced: "Advanced (ComfyUI)", describe: "Describe your scene",
    promptPh: "<Picture 1> walks through a rainy neon street at night...\n\nTip: point at your references with <Picture 1>, <Video 1>, <Audio 1>. Add a line starting with Audio: for sound.",
    promptTip: "English prompts work best. Name each reference with a tag (click a tag below to insert it). Add an \"Audio:\" line to describe sound and music.",
    refsTitle: "References", refsHint: "Show the model who or what to use. Each reference adds memory and time on a T4, so start with one or two. Videos are converted to 24 fps; anything longer than 15 s is cut.",
    pictures: "Pictures", videos: "Videos", audios: "Audios", addPictures: "Add pictures…", addVideo: "Add video…", addAudio: "Add audio…",
    soundtrack: "Use its soundtrack", detail: "Reference picture detail",
    d_match: "Match the video size (faster)", d_max: "Maximum detail (best likeness, can be several times slower)",
    detailHint: "\"Match\" shrinks pictures to the video's size. \"Maximum\" keeps up to 2048 px and can be much slower on a T4.",
    preparing: "Preparing", prepared: "Ready", trimmedNote: "cut to the first 15 s", limitImages: "You can add at most {n} pictures.",
    limitVideos: "You can add at most {n} videos.", limitAudios: "You can add at most {n} audios.", remove: "remove",
    frame: "Frame", landscape: "Landscape", portrait: "Portrait", square: "Square", quality: "Quality",
    draft: "Draft", standard: "Standard", high: "High", duration: "Duration", seconds: "s",
    secHint: "H3 builds videos in blocks of frames, so the real length is rounded up a little. At 15 s use Draft or Standard (High can run out of memory).",
    turbo: "Speed mode", turboHint: "Fewer steps = much faster. VDN is recommended; if it fails on a prompt, switch to Turbo. The 4-step Turbo quality is not verified.",
    t_off: "Off · 20 steps (slowest)", t_turbo8_768: "Turbo · 8 steps · 768p", t_turbo8: "Turbo · 8 steps",
    t_turbo4: "Turbo · 4 steps (experimental)", steps: "steps",
    t_vdn8: "VDN · 8 steps (recommended)",
    random: "Random", generate: "Generate video", cancel: "Cancel", ready: "Ready", download: "Download video",
    readyHint: "Add references, write a prompt and press Generate. The first run loads the models and takes longer.",
    ephemeral: "Videos live only in this Kaggle session. Download what you want to keep before the session ends.",
    license: "Powered by MiniMax H3 (read the model license before using or sharing results).",
    working: "Working", queued: "Waiting in the queue", sampling: "Generating", step: "Step",
    elapsed: "Elapsed", wait: "The first step is slow: models are being loaded.", done: "Done",
    stage_unet: "Loading the video model", stage_lora: "Loading Turbo", stage_vdn: "Applying VDN", stage_clip: "Reading your prompt",
    stage_vae_v: "Loading the video decoder", stage_vae_a: "Loading the audio decoder", stage_ref: "Reading your references",
    stage_cond: "Understanding the scene", stage_noise: "Preparing", stage_guider: "Preparing",
    stage_sampler: "Preparing", stage_sched: "Preparing", stage_sample: "Generating", stage_dec_v: "Decoding the video",
    stage_dec_a: "Decoding the audio", stage_video: "Assembling the video", stage_save: "Saving",
    err: "Something went wrong", uploadFail: "Upload failed", missing: "Model files are missing. Re-run the notebook download cell.",
    bd_title: "Where the time went", bd_load: "Loading models", bd_text: "Reading prompt / references", bd_sample: "Generating", bd_decode: "Decoding & saving",
    history: "This session", info: "{w}×{h}, {s} s, {st} steps", gpus: "GPUs: ",
    unauthorized: "Access key missing or expired. Open the full link from the notebook again."
  },
  tr: {
    tag: "· Referanstan video, Kaggle üzerinde MiniMax H3", advanced: "Gelişmiş (ComfyUI)", describe: "Sahneyi anlat",
    promptPh: "<Picture 1> yağmurlu bir neon sokakta yürüyor...\n\nİpucu: referansları <Picture 1>, <Video 1>, <Audio 1> ile göster. Ses için \"Audio:\" ile başlayan bir satır ekle.",
    promptTip: "İngilizce prompt daha iyi sonuç verir. Her referansı bir etiketle an (aşağıdaki etikete tıklayınca eklenir). Ses ve müziği anlatmak için \"Audio:\" satırı ekle.",
    refsTitle: "Referanslar", refsHint: "Modele kimi/neyi kullanacağını göster. T4'te her referans bellek ve süre ekler, bir ya da iki ile başla. Videolar 24 fps'e çevrilir; 15 sn'den uzunu kesilir.",
    pictures: "Görseller", videos: "Videolar", audios: "Sesler", addPictures: "Görsel ekle…", addVideo: "Video ekle…", addAudio: "Ses ekle…",
    soundtrack: "Sesini de kullan", detail: "Referans görsel ayrıntısı",
    d_match: "Video boyutuna uydur (daha hızlı)", d_max: "En yüksek ayrıntı (en iyi benzerlik, birkaç kat yavaş olabilir)",
    detailHint: "\"Uydur\" görselleri videonun boyutuna küçültür. \"En yüksek\" 2048 px'e kadar korur, T4'te çok daha yavaş olabilir.",
    preparing: "Hazırlanıyor", prepared: "Hazır", trimmedNote: "ilk 15 sn'ye kesildi", limitImages: "En fazla {n} görsel ekleyebilirsin.",
    limitVideos: "En fazla {n} video ekleyebilirsin.", limitAudios: "En fazla {n} ses ekleyebilirsin.", remove: "kaldır",
    frame: "Kadraj", landscape: "Yatay", portrait: "Dikey", square: "Kare", quality: "Kalite",
    draft: "Taslak", standard: "Standart", high: "Yüksek", duration: "Süre", seconds: "sn",
    secHint: "H3 videoyu kare bloklarıyla üretir, bu yüzden gerçek süre biraz yukarı yuvarlanır. 15 sn'de Taslak ya da Standart kullan (Yüksek belleği aşabilir).",
    turbo: "Hız modu", turboHint: "Daha az adım = çok daha hızlı. VDN önerilir; bir prompt'ta hata verirse Turbo'ya geç. 4 adımlı Turbo'nun kalitesi doğrulanmadı.",
    t_off: "Kapalı · 20 adım (en yavaş)", t_turbo8_768: "Turbo · 8 adım · 768p", t_turbo8: "Turbo · 8 adım",
    t_turbo4: "Turbo · 4 adım (deneysel)", steps: "adım",
    t_vdn8: "VDN · 8 adım (önerilen)",
    random: "Rastgele", generate: "Video üret", cancel: "İptal", ready: "Hazır", download: "Videoyu indir",
    readyHint: "Referans ekle, bir prompt yaz ve Üret'e bas. İlk çalıştırmada modeller yüklendiği için daha uzun sürer.",
    ephemeral: "Videolar sadece bu Kaggle oturumunda durur. Saklamak istediklerini oturum bitmeden indir.",
    license: "MiniMax H3 ile çalışır (kullanmadan ve sonuçları paylaşmadan önce model lisansını oku).",
    working: "Çalışıyor", queued: "Sırada bekliyor", sampling: "Üretiliyor", step: "Adım",
    elapsed: "Geçen süre", wait: "İlk adım yavaştır: modeller yükleniyor.", done: "Bitti",
    stage_unet: "Video modeli yükleniyor", stage_lora: "Turbo yükleniyor", stage_vdn: "VDN uygulanıyor", stage_clip: "Prompt okunuyor",
    stage_vae_v: "Video çözücü yükleniyor", stage_vae_a: "Ses çözücü yükleniyor", stage_ref: "Referanslar okunuyor",
    stage_cond: "Sahne anlaşılıyor", stage_noise: "Hazırlanıyor", stage_guider: "Hazırlanıyor",
    stage_sampler: "Hazırlanıyor", stage_sched: "Hazırlanıyor", stage_sample: "Üretiliyor", stage_dec_v: "Video çözülüyor",
    stage_dec_a: "Ses çözülüyor", stage_video: "Video birleştiriliyor", stage_save: "Kaydediliyor",
    err: "Bir şeyler ters gitti", uploadFail: "Yükleme başarısız", missing: "Model dosyaları eksik. Notebook'taki indirme hücresini yeniden çalıştır.",
    bd_title: "Zaman nereye gitti", bd_load: "Modeller yükleniyor", bd_text: "Prompt / referanslar okunuyor", bd_sample: "Üretim", bd_decode: "Çözme ve kaydetme",
    history: "Bu oturum", info: "{w}×{h}, {s} sn, {st} adım", gpus: "Ekran kartları: ",
    unauthorized: "Erişim anahtarı yok ya da süresi dolmuş. Notebook'taki tam bağlantıyı yeniden aç."
  }
};
const EXAMPLES = [
  ["Character", "<Picture 1> walks through a rainy neon street at night, glowing shop signs and puddles reflecting the lights, the camera follows slowly from behind, cinematic realism.\n\nAudio: rain, soft footsteps, quiet lo-fi music."],
  ["Motion", "Use <Video 1> as the motion reference: <Picture 1> performs the same street-dance moves in a sunlit plaza, handheld camera.\n\nAudio: upbeat street music."],
  ["Voice", "A close-up of <Picture 1> speaking calmly to the camera by a window in soft daylight. The voice sounds like <Audio 1>.\n\nAudio: the voice of <Audio 1>, quiet room tone."]
];

const $ = (s) => document.querySelector(s);
let lang = (() => { try { return localStorage.getItem("h3lang") === "tr" ? "tr" : "en"; } catch (e) { return "en"; } })();  // English by default
const t = (k) => (T[lang][k] ?? T.en[k] ?? k);
const clientId = (crypto.randomUUID ? crypto.randomUUID() : String(Math.random()).slice(2) + Date.now());

const state = { aspect: "landscape", quality: "draft", cfg: null, promptId: null,
  started: 0, timer: null, ws: null, stage: "", step: null, total: null, summary: null, finished: false, timeline: [],
  refs: { images: [], videos: [], audios: [] }, busyUpload: false };

// Graph nodes -> phase (for the "where the time went" summary) and -> label key.
const REF_NODE = /^(img|vid|vidc|vaud|aud)\d+$/;
const STATIC_PHASE = { unet: "load", lora: "load", vdn: "load", clip: "load", vae_v: "load", vae_a: "load",
  cond: "text", noise: "sample", guider: "sample", sampler: "sample", sched: "sample", sample: "sample",
  dec_v: "decode", dec_a: "decode", video: "decode", save: "decode" };
const phaseOf = (node) => REF_NODE.test(node) ? "text" : STATIC_PHASE[node];
const stageKey = (node) => REF_NODE.test(node) ? "stage_ref" : "stage_" + node;

function mark(node) { state.timeline.push({ node, t: Date.now() }); }

function renderBreakdown() {
  const box = $("#breakdown"); box.textContent = "";
  const tl = state.timeline; if (tl.length < 2) return;
  const totals = { load: 0, text: 0, sample: 0, decode: 0 };
  for (let i = 0; i < tl.length - 1; i++) { const phase = phaseOf(tl[i].node); if (phase) totals[phase] += tl[i + 1].t - tl[i].t; }
  const sum = Object.values(totals).reduce((a, b) => a + b, 0); if (!sum) return;
  const lines = ["load", "text", "sample", "decode"].map((k) => t("bd_" + k) + ": " + fmt(totals[k] / 1000) + " (" + Math.round(100 * totals[k] / sum) + "%)");
  const head = document.createElement("b"); head.textContent = t("bd_title") + " — " + fmt(sum / 1000);
  box.append(head, document.createElement("br"), document.createTextNode(lines.join(" · ")));
}

function applyLang() {
  document.documentElement.lang = lang;
  document.querySelectorAll("[data-i]").forEach((el) => { el.textContent = t(el.dataset.i); });
  document.querySelectorAll("[data-ip]").forEach((el) => { el.placeholder = t(el.dataset.ip); });
  renderExamples(); renderQuality(); renderSeconds(); renderHistoryLabel(); renderTurbo(); renderDetail(); renderRefs();
}

function renderTurbo() {
  const box = $("#turbo"); if (!state.cfg) return;
  const previous = box.value; box.textContent = "";
  const add = (value, label) => { const o = document.createElement("option"); o.value = value; o.textContent = label; box.appendChild(o); };
  state.cfg.turbo_options.forEach((o) => add(o.id, t("t_" + o.id)));
  add("off", t("t_off"));
  box.value = [...box.options].some((o) => o.value === previous) ? previous : box.options[0].value;
}

function renderDetail() {
  const box = $("#imageSize"); const previous = box.value || "match"; box.textContent = "";
  [["match", t("d_match")], ["max", t("d_max")]].forEach(([v, label]) => { const o = document.createElement("option"); o.value = v; o.textContent = label; box.appendChild(o); });
  box.value = previous;
}

function renderExamples() {
  const box = $("#examples"); box.textContent = "";
  EXAMPLES.forEach(([name, text]) => {
    const b = document.createElement("button"); b.type = "button"; b.textContent = name;
    b.onclick = () => { $("#prompt").value = text; };
    box.appendChild(b);
  });
}

function setSeg(box, value) { box.querySelectorAll("button").forEach((b) => b.classList.toggle("on", b.dataset.v === value)); }

function resolution(aspect, mp) {
  const r = { landscape: [16, 9], portrait: [9, 16], square: [1, 1] }[aspect];
  const scale = Math.sqrt(mp * 1024 * 1024 / (r[0] * r[1])), m = 32;
  return [Math.round(r[0] * scale / m) * m, Math.round(r[1] * scale / m) * m];
}

function renderQuality() {
  const box = $("#quality"); box.textContent = "";
  if (!state.cfg) return;
  Object.entries(state.cfg.quality).forEach(([name, q]) => {
    const [w, h] = resolution(state.aspect, q.megapixels);
    const b = document.createElement("button"); b.type = "button"; b.dataset.v = name;
    b.innerHTML = "<span></span><small></small>";
    b.firstChild.textContent = t(name); b.lastChild.textContent = w + "×" + h;
    b.onclick = () => { state.quality = name; setSeg(box, name); renderRes(); };
    box.appendChild(b);
  });
  if (!state.cfg.quality[state.quality]) state.quality = Object.keys(state.cfg.quality)[0];
  setSeg(box, state.quality); renderRes();
}

function renderRes() {
  if (!state.cfg) return;
  const q = state.cfg.quality[state.quality]; if (!q) return;
  const [w, h] = resolution(state.aspect, q.megapixels);
  $("#resText").textContent = w + " × " + h + " (" + q.megapixels + " MP)";
}

function renderSeconds() {
  const el = $("#seconds"); if (!state.cfg) return;
  $("#secLabel").textContent = el.value + " " + t("seconds");
  $("#secHint").textContent = t("secHint");
}

/* ---------------- references ---------------- */
const viewUrl = (name) => "/view?type=input&filename=" + encodeURIComponent(name) + "&subfolder=";

// <Audio j> numbering: the soundtracks of videos (when used) come first, in video order, then the stand-alone audios.
function assignAudioTags() {
  let j = 0;
  state.refs.videos.forEach((v) => { v.audioTag = (v.audio && v.soundtrack) ? ++j : null; });
  state.refs.audios.forEach((a) => { a.tag = ++j; });
}

function insertTag(text) {
  const box = $("#prompt"); const start = box.selectionStart ?? box.value.length, end = box.selectionEnd ?? start;
  box.value = box.value.slice(0, start) + text + box.value.slice(end);
  box.focus(); box.selectionStart = box.selectionEnd = start + text.length;
}

function tagButton(text, audio) {
  const b = document.createElement("button"); b.type = "button"; b.className = "tag" + (audio ? " audio" : ""); b.textContent = text;
  b.title = text; b.onclick = () => insertTag(text); return b;
}

function removeButton(list, index) {
  const x = document.createElement("button"); x.type = "button"; x.className = "ghost"; x.textContent = "×"; x.title = t("remove");
  x.onclick = () => { state.refs[list].splice(index, 1); renderRefs(); }; return x;
}

function renderRefs() {
  assignAudioTags();
  const lim = state.cfg ? state.cfg.references : { max_images: 9, max_videos: 3, max_audios: 3 };
  $("#countImages").textContent = "(" + state.refs.images.length + "/" + lim.max_images + ")";
  $("#countVideos").textContent = "(" + state.refs.videos.length + "/" + lim.max_videos + ")";
  $("#countAudios").textContent = "(" + state.refs.audios.length + "/" + lim.max_audios + ")";

  const imgBox = $("#listImages"); imgBox.textContent = "";
  state.refs.images.forEach((it, i) => {
    const row = document.createElement("div"); row.className = "refitem";
    const img = document.createElement("img"); img.src = viewUrl(it.name); img.alt = "";
    const meta = document.createElement("div"); meta.className = "meta"; meta.textContent = it.width + "×" + it.height;
    meta.append(document.createElement("br"), tagButton("<Picture " + (i + 1) + ">"));
    row.append(img, meta, removeButton("images", i)); imgBox.appendChild(row);
  });

  const vidBox = $("#listVideos"); vidBox.textContent = "";
  state.refs.videos.forEach((it, i) => {
    const row = document.createElement("div"); row.className = "refitem";
    const vid = document.createElement("video"); vid.src = viewUrl(it.name); vid.muted = true; vid.preload = "metadata"; vid.playsInline = true;
    const meta = document.createElement("div"); meta.className = "meta";
    meta.textContent = it.seconds + " s" + (it.trimmed ? " (" + t("trimmedNote") + ")" : "");
    meta.append(document.createElement("br"), tagButton("<Video " + (i + 1) + ">"));
    if (it.audioTag) meta.append(tagButton("<Audio " + it.audioTag + ">", true));
    if (it.audio) {
      const label = document.createElement("label"); label.style.display = "block"; label.style.marginTop = "4px";
      const cb = document.createElement("input"); cb.type = "checkbox"; cb.checked = !!it.soundtrack;
      cb.onchange = () => { it.soundtrack = cb.checked; renderRefs(); };
      label.append(cb, document.createTextNode(" " + t("soundtrack"))); meta.appendChild(label);
    }
    row.append(vid, meta, removeButton("videos", i)); vidBox.appendChild(row);
  });

  const audBox = $("#listAudios"); audBox.textContent = "";
  state.refs.audios.forEach((it, i) => {
    const row = document.createElement("div"); row.className = "refitem";
    const aud = document.createElement("audio"); aud.src = viewUrl(it.name); aud.controls = true; aud.preload = "metadata";
    const meta = document.createElement("div"); meta.className = "meta"; meta.textContent = it.seconds + " s" + (it.trimmed ? " (" + t("trimmedNote") + ")" : "");
    meta.append(document.createElement("br"), tagButton("<Audio " + it.tag + ">", true));
    row.append(aud, meta, removeButton("audios", i)); audBox.appendChild(row);
  });
}

function refStatus(kind, text) {
  const box = $("#refStatus"); box.textContent = text || ""; box.className = "refstatus " + (kind || "muted");
}

async function uploadRef(kind, file) {
  const lim = state.cfg.references, list = kind + "s";
  const max = { image: lim.max_images, video: lim.max_videos, audio: lim.max_audios }[kind];
  if (state.refs[list].length >= max) { refStatus("msg err", t({ image: "limitImages", video: "limitVideos", audio: "limitAudios" }[kind]).replace("{n}", max)); return; }
  refStatus("muted", t("preparing") + ": " + file.name + " …"); state.busyUpload = true; $("#go").disabled = true;
  try {
    const form = new FormData(); form.append("kind", kind); form.append("file", file);   // "kind" must come first
    const r = await fetch("/h3/api/reference", { method: "POST", body: form });
    if (r.status === 401) throw new Error(t("unauthorized"));
    const body = await r.json().catch(() => ({}));
    if (!r.ok) throw new Error(body.error || t("uploadFail"));
    if (kind === "video") body.soundtrack = !!body.audio;      // use the soundtrack by default when there is one
    state.refs[list].push(body); renderRefs(); refStatus("muted", t("prepared") + ": " + file.name);
  } catch (e) { refStatus("msg err", e.message || t("uploadFail")); }
  finally { state.busyUpload = false; if (!state.promptId || state.finished) $("#go").disabled = false; }
}

/* ---------------- generation ---------------- */
function notice(kind, text) {
  const box = $("#notice"); box.textContent = "";
  if (!text) return;
  const d = document.createElement("div"); d.className = "msg " + kind; d.textContent = text; box.appendChild(d);
}

function overlay(show, big, small, pct) {
  const o = $("#overlay"); o.style.display = show ? "flex" : "none"; if (!show) return;
  o.textContent = "";
  const a = document.createElement("div"); a.className = "big"; a.textContent = big || "";
  const b = document.createElement("div"); b.className = "muted"; b.textContent = small || "";
  o.append(a, b);
  if (pct != null) { const bar = document.createElement("div"); bar.className = "bar"; const i = document.createElement("i"); i.style.width = pct + "%"; bar.appendChild(i); o.appendChild(bar); }
}

function fmt(sec) { const m = Math.floor(sec / 60), s = Math.floor(sec % 60); return m + ":" + String(s).padStart(2, "0"); }

function tick() {
  if (state.finished) return;
  let big = t(state.stage ? stageKey(state.stage) : "working") || t("working");
  let small = t("elapsed") + " " + fmt((Date.now() - state.started) / 1000);
  let pct = null;
  if (state.stage === "sample" && state.step != null && state.total) {
    big = t("sampling") + " — " + t("step") + " " + state.step + "/" + state.total; pct = Math.round(100 * state.step / state.total);
  } else if (!state.stage) { big = t("queued"); }
  if (state.stage === "sample" && !state.step || state.stage === "unet" || state.stage === "clip") small += " · " + t("wait");
  overlay(true, big, small, pct);
}

function running(on) {
  $("#go").disabled = on; $("#stop").style.display = on ? "block" : "none";
  clearInterval(state.timer);
  if (on) { state.timer = setInterval(tick, 1000); tick(); }
}

async function jsonFetch(url, opts) {
  const r = await fetch(url, opts);
  if (r.status === 401) throw new Error(t("unauthorized"));
  let body = null; try { body = await r.json(); } catch (e) { /* not JSON */ }
  return { ok: r.ok, status: r.status, body };
}

function errorText(body) {
  if (!body) return t("err");
  if (typeof body.error === "string") return body.error;
  const e = body.error || {}; let text = e.message || t("err");
  const ne = body.node_errors || {};
  const first = Object.values(ne)[0];
  if (first && first.errors && first.errors[0]) text += " — " + (first.errors[0].message || "") + " " + (first.errors[0].details || "");
  return text;
}

function refsPayload() {
  return { images: state.refs.images.map((x) => x.name),
    videos: state.refs.videos.map((x) => ({ name: x.name, audio: x.audio || null, soundtrack: !!x.soundtrack })),
    audios: state.refs.audios.map((x) => x.name), image_size: $("#imageSize").value };
}

async function generate() {
  notice("", "");
  const prompt = $("#prompt").value.trim();
  if (!prompt) { notice("err", t("promptPh").split("\n")[0]); return; }
  const payload = { prompt, aspect: state.aspect, quality: state.quality, seconds: Number($("#seconds").value),
    turbo: $("#turbo").value, seed: Math.max(0, Math.floor(Number($("#seed").value) || 0)), refs: refsPayload() };
  try {
    const built = await jsonFetch("/h3/api/build", { method: "POST", headers: { "Content-Type": "application/json" }, body: JSON.stringify(payload) });
    if (!built.ok) { notice("err", errorText(built.body)); return; }
    state.summary = built.body.summary;
    const queued = await jsonFetch("/prompt", { method: "POST", headers: { "Content-Type": "application/json" },
      body: JSON.stringify({ prompt: built.body.prompt, client_id: clientId }) });
    if (!queued.ok) { notice("err", errorText(queued.body)); return; }
    state.promptId = queued.body.prompt_id; state.started = Date.now(); state.stage = ""; state.step = null; state.total = null; state.finished = false;
    state.timeline = []; mark("start"); $("#breakdown").textContent = "";
    $("#player").hidden = true; $("#tools").hidden = true; running(true); poll();
  } catch (e) { notice("err", e.message || t("err")); }
}

async function poll() {
  if (state.finished || !state.promptId) return;
  try {
    const h = await jsonFetch("/history/" + state.promptId);
    const entry = h.ok && h.body && h.body[state.promptId];
    if (entry && entry.status && (entry.status.completed || entry.status.status_str === "error")) { return finish(entry); }
  } catch (e) { /* keep waiting */ }
  setTimeout(poll, 3000);
}

function findVideo(entry) {
  for (const out of Object.values(entry.outputs || {})) {
    for (const list of Object.values(out || {})) {
      if (!Array.isArray(list)) continue;
      for (const f of list) { if (f && typeof f === "object" && /\.(mp4|webm|mov|mkv)$/i.test(f.filename || "")) return f; }
    }
  }
  return null;
}

function videoUrl(f) { return "/view?type=" + encodeURIComponent(f.type || "output") + "&filename=" + encodeURIComponent(f.filename) + "&subfolder=" + encodeURIComponent(f.subfolder || ""); }

function finish(entry) {
  if (state.finished) return;
  state.finished = true; running(false); mark("end"); renderBreakdown();
  const f = findVideo(entry);
  if (!f) {
    const msgs = (entry.status && entry.status.messages) || []; const bad = msgs.find((m) => m[0] === "execution_error");
    notice("err", bad ? (bad[1].exception_message || t("err")) : t("err")); overlay(true, t("err"), "");
    return;
  }
  showVideo(f, state.summary); notice("ok", t("done") + " — " + fmt((Date.now() - state.started) / 1000));
  loadHistory();
}

function showVideo(f, summary) {
  const url = videoUrl(f), p = $("#player");
  p.src = url; p.hidden = false; overlay(false); $("#tools").hidden = false;
  const a = $("#download"); a.href = url; a.download = f.filename;
  $("#resultInfo").textContent = summary ? t("info").replace("{w}", summary.width).replace("{h}", summary.height).replace("{s}", summary.seconds).replace("{st}", summary.steps) : f.filename;
  p.play().catch(() => {});
}

function connect() {
  const proto = location.protocol === "https:" ? "wss" : "ws";
  const ws = new WebSocket(proto + "://" + location.host + "/ws?clientId=" + clientId);
  state.ws = ws;
  ws.onmessage = (ev) => {
    if (typeof ev.data !== "string") return;
    let m; try { m = JSON.parse(ev.data); } catch (e) { return; }
    const d = m.data || {};
    if (d.prompt_id && state.promptId && d.prompt_id !== state.promptId) return;
    if (m.type === "executing" && d.node) { state.stage = d.node; mark(d.node); if (d.node !== "sample") { state.step = null; } }
    else if (m.type === "progress" && d.max) { state.step = d.value; state.total = d.max; state.stage = "sample"; }
    else if (m.type === "execution_error") { state.finished = true; running(false); mark("end"); renderBreakdown(); notice("err", d.exception_message || t("err")); overlay(true, t("err"), ""); }
    else if (m.type === "execution_success" || (m.type === "executing" && !d.node && state.promptId)) { poll(); }
  };
  ws.onclose = () => { setTimeout(connect, 2500); };
}

async function loadHistory() {
  try {
    const h = await jsonFetch("/history?max_items=30");
    if (!h.ok || !h.body) return;
    const items = [];
    Object.entries(h.body).forEach(([id, e]) => { const f = findVideo(e); if (f && /^H3Studio/.test(f.filename)) items.push(f); });
    const box = $("#history"); box.textContent = "";
    if (items.length) { const title = document.createElement("div"); title.className = "muted"; title.dataset.i = "history"; title.textContent = t("history"); box.appendChild(title); }
    items.slice(-12).reverse().forEach((f) => {
      const b = document.createElement("button"); b.type = "button"; b.textContent = f.filename;
      b.onclick = () => { showVideo(f, null); }; box.appendChild(b);
    });
  } catch (e) { /* ignore */ }
}
function renderHistoryLabel() { const el = document.querySelector("#history [data-i=history]"); if (el) el.textContent = t("history"); }

async function init() {
  document.querySelectorAll("#aspect button").forEach((b) => { b.onclick = () => { state.aspect = b.dataset.v; setSeg($("#aspect"), state.aspect); renderQuality(); }; });
  $("#langBtn").onclick = () => { lang = lang === "tr" ? "en" : "tr"; try { localStorage.setItem("h3lang", lang); } catch (e) { /* storage blocked */ } applyLang(); };
  $("#dice").onclick = () => { $("#seed").value = Math.floor(Math.random() * 2 ** 31); };
  $("#seconds").oninput = renderSeconds;
  $("#go").onclick = generate;
  $("#stop").onclick = async () => { try { await fetch("/interrupt", { method: "POST" }); } catch (e) { /* ignore */ } };
  [["image", "#addImage", "#fileImage"], ["video", "#addVideo", "#fileVideo"], ["audio", "#addAudio", "#fileAudio"]].forEach(([kind, button, input]) => {
    $(button).onclick = () => $(input).click();
    $(input).onchange = async (e) => { const files = [...e.target.files]; e.target.value = ""; for (const file of files) { await uploadRef(kind, file); } };
  });
  $("#seed").value = Math.floor(Math.random() * 2 ** 31);
  applyLang();
  try {
    const c = await jsonFetch("/h3/api/config");
    if (!c.ok) throw new Error(t("unauthorized"));
    state.cfg = c.body;
    const s = $("#seconds"); s.min = c.body.min_seconds; s.max = c.body.max_seconds; s.step = 0.5; s.value = Math.min(5, c.body.max_seconds);
    $("#gpuInfo").textContent = t("gpus") + (c.body.gpus.join(" + ") || "—");
    if (c.body.missing.length) notice("warn", t("missing") + " (" + c.body.missing.join(", ") + ")");
    applyLang();
  } catch (e) { notice("err", e.message); }
  connect(); loadHistory();
}
init();
</script>
</body>
</html>
''', encoding="utf-8")
print("written:", PLUGIN_DIR / "studio.html")
```

## Step 5: Download the models

About 50 GB. Every file is verified by **size + SHA256**; if it does not match, it is deleted and the notebook stops.

```python
# Step 5 - Download the models and verify them with SHA256 (resumes if interrupted)
import hashlib
import os
import time
from pathlib import Path

from huggingface_hub import hf_hub_download

os.environ["HF_HOME"] = str(WORK_ROOT / "hf_cache")
os.environ["HF_HUB_ENABLE_HF_TRANSFER"] = "0"
MODELS_DIR = COMFY_HOME / "models"

def sha256_of(path, label):
    h, total, t0 = hashlib.sha256(), path.stat().st_size, time.time()
    with path.open("rb") as f:
        for block in iter(lambda: f.read(8 * 1024 * 1024), b""):
            h.update(block)
    print(f"    SHA256 computed ({total / GIB:.1f} GiB, {time.time() - t0:.0f} s)")
    return h.hexdigest()

def fetch(repo, rel, size, digest, folder=None):
    """Downloads repo/rel into models/<folder>/; stops if the size or SHA256 does not match."""
    target = MODELS_DIR / (rel if folder is None else f"{folder}/{Path(rel).name}")
    target.parent.mkdir(parents=True, exist_ok=True)
    if target.exists() and target.stat().st_size == size and sha256_of(target, rel) == digest:
        print(f"  ✓ Already verified: {target.name}")
        return
    if target.exists():
        print(f"  Re-downloading {target.name}: size/hash mismatch, deleting it")
        target.unlink()
    print(f"  Downloading {repo} -> {target.name} ({size / GIB:.1f} GiB)")
    got = Path(hf_hub_download(repo_id=repo, filename=rel if folder is None else Path(rel).name,
                               local_dir=target.parent if folder else MODELS_DIR))
    if got.resolve() != target.resolve():
        got.replace(target)
    if target.stat().st_size != size:
        raise RuntimeError(f"{target.name}: size {target.stat().st_size} != {size}")
    if sha256_of(target, rel) != digest:
        target.unlink()
        raise RuntimeError(f"{target.name}: SHA256 mismatch, file deleted. Try again.")
    print(f"  ✓ Verified: {target.name}")

for repo, rel, size, digest in MODEL_FILES:
    fetch(repo, rel, size, digest)
for key in TURBO_DOWNLOAD:
    rel, size, digest = TURBO_FILES[key]
    fetch(TURBO_REPO, rel, size, digest, folder="loras")

print("\n  ✓ All model files are ready and verified")
```

## Step 5b: VDN-H3 patch (on by default)

Installs the ComfyUI port of **VDN-H3**, an 8-step distilled, hybrid-attention version of MiniMax H3 (third-party node at a pinned commit; the VDN stage and the grid file are SHA256-verified; ~3.5 GB). It appears **first and selected** in the page's **Speed mode** list. To turn it off, set `INSTALL_VDN_EXPERIMENT = False` in Step 1. If the install fails the notebook **does not stop**; it continues with Turbo. The node was written for the RTX 5090 and has not yet been tried on a T4 with this reference model: if it errors on your references, pick a Turbo mode in the page.

```python
# Step 5b - VDN-H3 patch (runs when INSTALL_VDN_EXPERIMENT = True in Step 1; on by default).
# It installs: (1) the VDN node (PINNED commit), (2) the VDN stage (SHA256-verified),
# (3) the AdaLN grid file (SHA256-verified). No pip packages are installed (NVIDIA CuTe is optional and skipped).
# The node's code only opens read-only safetensors files; it has no network access, subprocess, eval/exec (code review: README).
# If anything goes wrong the notebook does NOT stop: the VDN mode stays off and the Turbo modes keep working.
import hashlib
import shutil
import subprocess
import urllib.request


def install_vdn():
    from huggingface_hub import hf_hub_download

    def verified(path, size, digest):
        """Does the file exist with the expected size and SHA256?"""
        if not path.is_file() or path.stat().st_size != size:
            return False
        h = hashlib.sha256()
        with path.open("rb") as f:
            for block in iter(lambda: f.read(8 * 1024 * 1024), b""):
                h.update(block)
        return h.hexdigest() == digest

    # 1) The node: a single pinned commit
    node_dir = COMFY_HOME / "custom_nodes" / "ComfyUI-VDN-H3-Fast"
    if not (node_dir / ".git").exists():
        node_dir.mkdir(parents=True, exist_ok=True)
        sh(["git", "init", "--quiet", node_dir])
        sh(["git", "-C", node_dir, "fetch", "--depth", "1", VDN_NODE_REPO, VDN_NODE_COMMIT])
        sh(["git", "-C", node_dir, "checkout", "--quiet", "--detach", "FETCH_HEAD"])
    head = subprocess.check_output(["git", "-C", str(node_dir), "rev-parse", "HEAD"], text=True).strip()
    if head != VDN_NODE_COMMIT:
        raise RuntimeError(f"VDN node commit mismatch: {head} != {VDN_NODE_COMMIT}")
    print("  VDN node verified:", head[:12])

    # 2) The VDN stage (keep the folder layout: model_spec.json, linear_branch/, adapters/)
    stage_dir = MODELS_DIR / "vdn" / VDN_STAGE_DIR
    for rel, size, digest in VDN_STAGE_FILES:
        target = stage_dir / rel
        if verified(target, size, digest):
            print(f"  ✓ Already verified: {rel}")
            continue
        target.parent.mkdir(parents=True, exist_ok=True)
        if target.exists():
            target.unlink()
        print(f"  Downloading {VDN_STAGE_REPO}@{VDN_STAGE_REVISION[:8]} -> {rel} ({size / GIB:.2f} GiB)")
        hf_hub_download(repo_id=VDN_STAGE_REPO, revision=VDN_STAGE_REVISION, filename=rel, local_dir=stage_dir)
        if not verified(target, size, digest):
            target.unlink(missing_ok=True)
            raise RuntimeError(f"{rel}: size/SHA256 mismatch, file deleted.")
        print(f"  ✓ Verified: {rel}")

    # 3) The AdaLN grid
    grid = MODELS_DIR / "vdn" / "h3_silu_temb_grid.safetensors"
    if not verified(grid, VDN_GRID_SIZE, VDN_GRID_SHA256):
        grid.parent.mkdir(parents=True, exist_ok=True)
        print("  Downloading the AdaLN grid (5.5 MB)...")
        urllib.request.urlretrieve(VDN_GRID_URL, str(grid))
        if not verified(grid, VDN_GRID_SIZE, VDN_GRID_SHA256):
            grid.unlink(missing_ok=True)
            raise RuntimeError("Grid file SHA256 mismatch, deleted.")
    print("  ✓ Verified: h3_silu_temb_grid.safetensors")


VDN_READY = False
if not INSTALL_VDN_EXPERIMENT:
    print("VDN is off (INSTALL_VDN_EXPERIMENT=False): skipped. The Turbo modes will be used.")
else:
    try:
        install_vdn()
        VDN_READY = True
        print("\n  ✓ VDN is ready. In the page, Speed mode -> 'VDN - 8 steps (recommended)' will be first in the list.")
        print("  Note: the node was written for the RTX 5090; on T4 it is not guaranteed for every setting. If it errors, switch to a Turbo mode.")
    except Exception as exc:  # VDN is optional: if it cannot be installed we carry on with Turbo
        print(f"\n  ⚠ VDN could not be installed ({type(exc).__name__}: {exc}).")
        print("    The VDN mode stays off; the Turbo modes work normally. (The notebook keeps going.)")
        shutil.rmtree(COMFY_HOME / "custom_nodes" / "ComfyUI-VDN-H3-Fast", ignore_errors=True)  # remove a half-installed node
```

```python
# Step 6 - Launch H3 Studio and open a password-protected public link
# While this cell runs, Studio is open. Stop it with the square button (■) to shut everything down.
# Order: ComfyUI starts -> it tests the password protection BY ITSELF before any tunnel exists -> only if that passes, the tunnel opens.
import hashlib
import os
import re
import secrets
import signal
import subprocess
import sys
import time
import urllib.error
import urllib.request

import requests
from IPython.display import HTML, clear_output, display

BASE_URL = f"http://127.0.0.1:{PORT}"
ACCESS_KEY = secrets.token_urlsafe(24)            # new random key on every run
for old in ("server.pid", "tunnel.pid"):          # if this cell is re-run, stop the old processes
    pf = WORK_ROOT / old
    if pf.exists():
        try:
            os.kill(int(pf.read_text()), signal.SIGTERM)
            time.sleep(2)
        except Exception:
            pass
        pf.unlink(missing_ok=True)

env = os.environ.copy()
env.update(PYTORCH_CUDA_ALLOC_CONF="expandable_segments:True", H3_STUDIO_KEY=ACCESS_KEY,
           H3_STUDIO_MAX_MP=str(MAX_MEGAPIXELS), H3_STUDIO_MAX_SECONDS=str(MAX_SECONDS),
           HF_HOME=str(WORK_ROOT / "hf_cache"))
args = [sys.executable, "main.py", "--listen", "127.0.0.1", "--port", str(PORT), "--lowvram",
        "--preview-method", "none", "--disable-api-nodes", "--disable-auto-launch",
        "--output-directory", str(GEN_DIR), "--temp-directory", str(TMP_DIR)]
helptext = subprocess.run([sys.executable, "main.py", "--help"], cwd=COMFY_HOME, env=env,
                          capture_output=True, text=True).stdout
if "--reserve-vram" in helptext:
    args += ["--reserve-vram", "0.8"]
print("$", " ".join(args))          # the key is NOT on the command line, it is in an environment variable
log_fp = open(LOG_FILE, "w", encoding="utf-8")
SERVER = subprocess.Popen(args, cwd=COMFY_HOME, env=env, stdout=log_fp, stderr=subprocess.STDOUT)
(WORK_ROOT / "server.pid").write_text(str(SERVER.pid))

def status(path, headers=None):
    request = urllib.request.Request(BASE_URL + path, headers=headers or {})
    try:
        return urllib.request.urlopen(request, timeout=10).status
    except urllib.error.HTTPError as error:
        return error.code

t0 = time.time()
while True:
    if SERVER.poll() is not None:
        raise RuntimeError("ComfyUI exited right after starting:\n" + log_tail(100))
    try:
        if status("/system_stats") in (200, 401):
            break
    except Exception:
        pass
    if time.time() - t0 > 300:
        raise RuntimeError("ComfyUI was not ready after 5 minutes:\n" + log_tail(100))
    time.sleep(2)
print(f"ComfyUI is ready ({time.time() - t0:.0f} s)")

# ---- Security self-test: NOTHING may open without the key ----
for path in ("/", "/h3", "/h3/api/config", "/system_stats", "/object_info", "/queue", "/history", "/extensions"):
    code = status(path)
    if code != 401:
        SERVER.terminate()
        raise RuntimeError(f"SECURITY TEST FAILED: {path} without the key returned HTTP {code} (expected 401). "
                           "The tunnel was NOT opened. The plugin may have failed to load:\n" + log_tail(60))
if status("/h3/api/config", {"X-H3-Key": ACCESS_KEY}) != 200:
    SERVER.terminate()
    raise RuntimeError("A request with the key was rejected; the plugin is not working:\n" + log_tail(60))
cfg = requests.get(BASE_URL + "/h3/api/config", headers={"X-H3-Key": ACCESS_KEY}, timeout=10).json()
if cfg["missing"]:
    SERVER.terminate()
    raise RuntimeError("Model files are missing: " + ", ".join(cfg["missing"]))
print("✓ Security test passed: requests without the key are rejected with 401.")
print(f"✓ Plugin ready - {len(cfg['gpus'])} GPU(s): {', '.join(cfg['gpus'])}")
print("✓ Speed modes available:", ", ".join(o["id"] for o in cfg["turbo_options"]) or "none (only the slow 20-step mode)")

# ---- cloudflared (pinned version, SHA256-verified) ----
cf_bin = WORK_ROOT / "cloudflared"
if not cf_bin.exists() or hashlib.sha256(cf_bin.read_bytes()).hexdigest() != CLOUDFLARED_SHA256:
    print(f"Downloading cloudflared {CLOUDFLARED_VERSION}...")
    urllib.request.urlretrieve(CLOUDFLARED_URL, str(cf_bin))
    if hashlib.sha256(cf_bin.read_bytes()).hexdigest() != CLOUDFLARED_SHA256:
        cf_bin.unlink()
        SERVER.terminate()
        raise RuntimeError("cloudflared SHA256 mismatch; the download is not trustworthy, stopped.")
    os.chmod(cf_bin, 0o755)
tunnel_log = WORK_ROOT / "cloudflared.log"
TUNNEL = subprocess.Popen([str(cf_bin), "--no-autoupdate", "tunnel", "--url", BASE_URL],
                          stdout=open(tunnel_log, "w", encoding="utf-8"), stderr=subprocess.STDOUT)
(WORK_ROOT / "tunnel.pid").write_text(str(TUNNEL.pid))
PUBLIC_URL, t0 = None, time.time()
while time.time() - t0 < 90 and not PUBLIC_URL:
    if TUNNEL.poll() is not None:
        SERVER.terminate()
        raise RuntimeError("cloudflared exited:\n" + tunnel_log.read_text(errors="replace")[-600:])
    time.sleep(2)
    found = re.findall(r"https://[\w-]+\.trycloudflare\.com", tunnel_log.read_text(errors="replace"))
    PUBLIC_URL = found[0] if found else None
if not PUBLIC_URL:
    SERVER.terminate()
    TUNNEL.terminate()
    raise RuntimeError("cloudflared did not give an address within 90 s.")

STUDIO_LINK = f"{PUBLIC_URL}/h3?key={ACCESS_KEY}"
display(HTML(f"""
<div style="border:2px solid #3ac97a;border-left:12px solid #2bff88;border-radius:12px;padding:18px 24px;margin:14px 0;
            background:rgba(43,217,130,.10);font-family:system-ui,'Segoe UI',sans-serif;">
  <div style="font-size:20px;font-weight:800;">🎬 Studio is ready</div>
  <div style="margin-top:8px;font-size:16px;"><a href="{STUDIO_LINK}" target="_blank"><b>Open H3 Studio →</b></a></div>
  <div style="margin-top:8px;opacity:.8;">This link <b>contains a password</b>: do not share it and do not publish this notebook's output.<br>
  When you are done, <b>download your videos</b> from the page first, then stop this cell with ■.</div>
</div>"""))

print("Studio is open. (Stop this cell with ■ to shut it down.)")
try:
    while SERVER.poll() is None and TUNNEL.poll() is None:
        time.sleep(30)
    print("\nA process exited unexpectedly. End of the ComfyUI log:")
    print(log_tail(80))
except KeyboardInterrupt:
    print("\nShutting down...")
finally:
    for proc in (TUNNEL, SERVER):
        if proc.poll() is None:
            proc.terminate()
            try:
                proc.wait(timeout=10)
            except Exception:
                proc.kill()
    for old in ("server.pid", "tunnel.pid"):
        (WORK_ROOT / old).unlink(missing_ok=True)
    log_fp.flush()
    clear_output(wait=False)      # remove the password-protected link from the output
    print("Stopped. The link is no longer valid. Videos are in:", GEN_DIR)
    print("To keep them, run the next cell (Step 7).")
```
